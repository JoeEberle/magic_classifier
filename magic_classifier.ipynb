```python
import seaborn as sns
import matplotlib.pyplot as plt

# Plot
ax = sns.countplot(x='species', data=df_penguins)

# Add counts above the bars
for p in ax.patches:
    count = int(p.get_height())
    ax.annotate(count,                     # Text to show
                (p.get_x() + p.get_width() / 2., count),  # x, y position
                ha='center', va='bottom',  # Align center/bottom
                fontsize=10)

plt.tight_layout()
plt.show()
```
